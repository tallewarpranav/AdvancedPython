```python
# Random Variable
# Random + Variable
# Random: Outcome is uncertain
# Variable: Value can change.
# Therefore, a Random Variable: A variable whose value depends on the outcome of a random event.
# In simple terms, It is a number assigned to uncertain outcomes.

# Example:
# Suppose:
# You track:
# Number of orders received in one hour

# Possible values:
# 10
# 20
# 35
# 50

# You cannot predict exact number beforehand.

# Hence:
# This is a random variable.

# Real-life example:
# Suppose:
# You monitor:
# Number of fraud transactions per day

# You don't know exact value in advance.
# Possible outcomes vary randomly.

# So:
# Fraud count is a random variable.

# Another example:
# Suppose factory tracks:
# Number of defective products produced daily

# Some days:
# 2 defects
# 5 defects
# 0 defects

# Outcome varies randomly.

# Hence:
# Defect count is random variable.

# Why Random Variables Are Important?
# Because probability and statistics are built around them.

# They help answer:
# - What may happen?
# - How likely?
# - What is expected?
# - What is risk?
```

```python
# Two Main Types of Random Variables
# 1. Discrete Random Variable: Takes specific values (e.g., number of heads in coin tosses) i.e. Countable outcomes
# 2. Continuous Random Variable: Takes any value within a range (e.g., height of people) i.e. Uncountable outcomes

# 1. Discrete Random Variable
# Takes:
# Countable values
```

| Example            | Why Discrete? |
| ------------------ | ------------- |
| Number of students | Countable     |
| Number of defects  | Countable     |
| Number of calls    | Countable     |


```python
# Example — Dice Roll
# Suppose:
# X = Number appearing on dice

# Possible values:
# 1,2,3,4,5,6

# Finite countable values.

# So:
# Discrete Random Variable


# 2. Continuous Random Variable
# Takes:
# Measurable values
# Can contain decimals.
```

| Example       | Why Continuous?         |
| ------------- | ----------------------- |
| Temperature   | Infinite decimal values |
| Height        | Measurable              |
| Delivery time | Continuous scale        |


```python
# Example — Delivery Time
# Suppose:
# X = Delivery time

# Possible values:
# 20.5 min
# 21.2 min
# 18.9 min

# Continuous measurable values.

# So:
# Continuous Random Variable

# Real-life example: Weather Example
# Suppose:
# X = Tomorrow’s rainfall

# You don't know exact value beforehand.

# Could be:
# 0 mm
# 10 mm
# 50 mm

# This is a random variable.
```

```python
# Probability Distribution

# Suppose you ask:
# “What are the possible outcomes of a random event, and how likely is each outcome?”

# That complete information is called:
# Probability Distribution

# Probability distribution tells us:
# All possible values a random variable can take and the probability of each value.

# Real-life example - Dice Roll
# Suppose:
# X = Number appearing on dice

# Possible outcomes:
# 1, 2, 3, 4, 5, 6

# Each outcome has probability:
# 1/6

# This complete mapping becomes:
# Probability Distribution
```

| 2           | 1/6              |
| 3           | 1/6              |
| 4           | 1/6              |
| 5           | 1/6              |

```python
# Expected Value:

# Expected value means:
# The long-run average outcome of a random process.

# In simple terms, It tells us what value we should expect on average if an experiment is repeated many times.

# Expected value is NOT always an actually occurring value.
# It is:
# Theoretical average outcome.

# Real-Life Example — Swiggy Orders
# Suppose probabilities of orders per hour are:
```

| Orders | Probability |
| ------ | ----------- |
| 10     | 0.2         |
| 20     | 0.5         |
| 30     | 0.3         |


```python
# Expected Orders
# E(X)=10(0.2)+20(0.5)+30(0.3)

# Solve
# 2+10+9=21

# Final Answer
# Expected orders per hour = 21

# Real-Life Business Meaning:
# Company can plan resources assuming:
# ~21 orders/hour on average.
```

```python
# Real-Life Banking Example
# Suppose:
# Loan default probability exists.

# Banks calculate:
# Expected loss
# using expected value.
```

| Loss Amount | Probability |
| ----------- | ----------- |
| ₹0          | 0.8         |
| ₹1 lakh     | 0.2         |


```python
# Expected loss:
# 0(0.8)+100000(0.2)=20000

# Interpretation
# Bank expects:
# ₹20,000 average loss per customer.
```

```python
# Real-Life Manufacturing Example
# Suppose machine failure costs:
# ₹50,000

# Failure probability:
# 0.1

# Expected maintenance cost:
# 50000(0.1)=5000

# Meaning
# Company should plan roughly:
# ₹5000 expected maintenance risk.
```

```python
# Discrete Probability Distributions:

# These distributions are used when:
# Outcomes are countable/discrete.

# Examples:
# - Number of defects
# - Number of calls
# - Number of customers
# - Success/failure events
```

| Distribution | Used For                                  |
| ------------ | ----------------------------------------- |
| Bernoulli    | Single success/failure event              |
| Binomial     | Multiple repeated success/failure events  |
| Poisson      | Count of events occurring over time/space |


```python
# 1. Bernoulli Distribution
# Simplest Probability Distribution

# What Is Bernoulli Distribution?
# Used when:
# Only TWO possible outcomes exist.

# Usually:
```

| Outcome | Meaning |
| ------- | ------- |
| 1       | Success |
| 0       | Failure |


```python
# In simple terms,
# Bernoulli answers:
# “Did success happen or not?”
```

### Real-Life Examples

| Scenario             | Success/Failure   |
| -------------------- | ----------------- |
| Coin toss            | Head/Tail         |
| Email spam detection | Spam/Not Spam     |
| Loan approval        | Approved/Rejected |
| Machine check        | Pass/Fail         |


```python
# Example — Coin Toss
# Suppose:
# Head = 1
# Tail = 0

# Probability of head:
# p

# Probability of tail:
# 1-p

# This becomes:
# Bernoulli Distribution
```

### Important Properties

| Property | Value        |
| -------- | ------------ |
| Outcomes | Only 0 or 1  |
| Trials   | Single trial |
| Mean     | p            |
| Variance | p(1-p)       |


```python
# 2. Binomial Distribution
# Now suppose:
# We repeat Bernoulli experiment multiple times.
# That becomes:
# Binomial Distribution

# What Is Binomial Distribution?
# Used when:
# - Fixed number of trials exist
# Each trial has:
# - Success or failure
# - Probability remains constant

# Binomial answers:
# “How many successes occur in multiple trials?”
```

```python
# 3. Poisson Distribution

# What Is Poisson Distribution?
# Used when counting:
# Number of events occurring in fixed interval.

# Poisson answers:
# “How many times an event may occur in a given time/space?”
```
