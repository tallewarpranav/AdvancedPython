Below is a **cheat sheet-style reference** of all major file formats supported by **Pandas** for reading data, along with **example syntax** for each.

---

## ✅ 1. CSV (Comma-Separated Values)

```python
import pandas as pd

df = pd.read_csv('data.csv')  # Basic
df = pd.read_csv('data.csv', delimiter=',', encoding='utf-8')  # With options
```

---

## ✅ 2. Excel (`.xls`, `.xlsx`)

```python
df = pd.read_excel('data.xlsx', sheet_name='Sheet1')  # Read specific sheet
```

> Requires: `openpyxl` for `.xlsx`, `xlrd` for `.xls`

---

## ✅ 3. JSON (JavaScript Object Notation)

```python
df = pd.read_json('data.json')  # Assumes records are JSON objects
```

> If using a JSON string or URL:

```python
df = pd.read_json(json_string_or_url, orient='records')
```

---

## ✅ 4. Parquet (Apache columnar format)

```python
df = pd.read_parquet('data.parquet')
```

> Requires: `pyarrow` or `fastparquet`

---

## ✅ 5. SQL (Databases)

```python
import sqlite3

conn = sqlite3.connect('my_database.db')
df = pd.read_sql('SELECT * FROM my_table', conn)
```

> For other DBs like PostgreSQL, use `sqlalchemy` or `psycopg2`.

---

## ✅ 6. HTML (Web Tables)

```python
dfs = pd.read_html('https://example.com/page-with-tables')  # Returns list of DataFrames
```

> Requires: `lxml` or `html5lib`

---

## ✅ 7. Clipboard

```python
df = pd.read_clipboard()  # Paste tabular data copied from Excel or web
```

---

## ✅ 8. ORC (Optimized Row Columnar)

```python
df = pd.read_orc('data.orc')
```

> Requires: `pyarrow`

---

## ✅ 9. Feather

```python
df = pd.read_feather('data.feather')
```

> Fast binary format (good for pipeline stages)

---

## ✅ 10. Pickle (Python serialized object)

```python
df = pd.read_pickle('data.pkl')
```

---

## ✅ 11. XML

```python
df = pd.read_xml('data.xml')
```

> Requires: `lxml`

---

## ✅ 12. Google Sheets (via URL)

```python
sheet_url = 'https://docs.google.com/spreadsheets/d/xyz123/export?format=csv'
df = pd.read_csv(sheet_url)
```

> Ensure the sheet is **shared publicly or accessibly via link**

---

## ✅ 13. Reading from ZIP File (compressed)

```python
df = pd.read_csv('data.zip', compression='zip')
```

> Works similarly for gzip, bz2, xz, etc.

---

## ✅ 14. Reading from S3 / HTTPS / GCS / Azure Blob (Cloud URLs)

```python
df = pd.read_csv('https://bucket.s3.amazonaws.com/data.csv')
df = pd.read_parquet('gs://your-bucket/data.parquet')  # if gcsfs installed
```

> Requires: credentials + cloud FS library (e.g., `s3fs`, `gcsfs`, `adlfs`)

````python
# Indexers in Pandas are used to access and manipulate data in a DataFrame or Series. They allow you to select specific rows and columns based on labels or positions. There are two main types of indexers in Pandas: .loc and .iloc.
# loc is used for label-based indexing, which means you can select rows and columns based on their labels (index names). It allows you to access data using the index labels and column names.
# iloc is used for position-based indexing, which means you can select rows and columns based on their integer positions. It allows you to access data using the integer positions of the rows and columns in the DataFrame or Series.

print(data.loc["a"])
print(data.iloc[0])
print(".."*50)
print(data.loc["b"])
print(data.iloc[1])

print("=="*50)
print(data.iloc[0:2])
print(".."*50)
print(data.loc["a" : "b"]) # SI= "a", EI= "b", both are included in the output
````
