````python
# Introduction to Pandas

# Pandas
# Pandas is an important library for data manipulation, wrangling, and analysis.
# What is Data Manipulation?
# Eg: If we had a price column in a dataset, we might want to manipulate the data by converting the prices from one currency to another. This is an example of data manipulation.
# What is Data Wrangling?
# Data wrangling is the process of cleaning and transforming raw data into a format that is suitable for analysis. This can involve tasks such as handling missing values, removing duplicates, and converting data types.
# Eg: If we had a price column in a dataset with values in different formats (e.g., "$10", "10 USD", "10"), we might want to wrangle the data by standardizing the format of the prices. This is an example of data wrangling.
# Eg: If we had a price column in a dataset with some missing values, we might want to wrangle the data by filling in the missing values with the average price. This is another example of data wrangling.
# What is Data Analysis?
# Data analysis is the process of examining, cleaning, transforming, and modeling data to discover useful information, draw conclusions, and support decision-making.
# Eg: If we had a price column in a dataset, we might want to analyze the data by calculating the average price, finding the maximum and minimum prices, and identifying any trends or patterns in the data. This is an example of data analysis.

# Pandas: Pan - Panel Data, das - Data Analysis System
# It is built on top of NumPy and is designed to work seamlessly with it.
# It is also the most widely used library for data analysis in Python.
# It is used in a variety of fields, including finance, economics, statistics, and data science.
# It is also used in machine learning and deep learning.

# Data Structures of Pandas
# Pandas provides two main data structures: Series and DataFrame.
# 1. Series:
#    - A one-dimensional labeled array capable of holding any data type.
#    - It is similar to a list or a dictionary, but it has an index that allows for easy access to the data.
#    - It can be created from a list, dictionary, or NumPy array.
# 2. DataFrame:
#    - A two-dimensional labeled data structure with columns of potentially different types.
#    - It is similar to a spreadsheet or SQL table, or a dictionary of Series objects.
#    - It can be created from a dictionary of Series, a list of dictionaries, or a NumPy array.
#    - It is the most commonly used data structure in Pandas and is used for most data analysis tasks.
#    - It is also the most powerful data structure in Pandas and is used for most data manipulation tasks.
#    - It is used for data wrangling, data cleaning, and data analysis.
#    - It is used for data visualization and data exploration.
#    - It is used for data preprocessing and feature engineering.
#    - It is used for data modeling and machine learning.
#    - It is used for data storage and data retrieval.
#    - It is used for data transformation and data aggregation.
#    - It is used for data merging and data joining.
#    - It is used for data reshaping and data pivoting.
#    - It is used for data filtering and data selection.
#    - It is used for data grouping and data aggregation.
#    - It is used for data sorting and data ordering.
#    - It is used for data indexing and data slicing.
#    - It is used for data concatenation and data appending.
#    - It is used for data splitting and data merging.
````

Below is a **cheat sheet-style reference** of all major file formats supported by **Pandas** for reading data, along with **example syntax** for each.

---

## ✅ 1. CSV (Comma-Separated Values)

```python
import pandas as pd

df = pd.read_csv('data.csv')  # Basic
df = pd.read_csv('data.csv', delimiter=',', encoding='utf-8')  # With options
```

---

## ✅ 2. Excel (`.xls`, `.xlsx`)

```python
df = pd.read_excel('data.xlsx', sheet_name='Sheet1')  # Read specific sheet
```

> Requires: `openpyxl` for `.xlsx`, `xlrd` for `.xls`

---

## ✅ 3. JSON (JavaScript Object Notation)

```python
df = pd.read_json('data.json')  # Assumes records are JSON objects
```

> If using a JSON string or URL:

```python
df = pd.read_json(json_string_or_url, orient='records')
```

---

## ✅ 4. Parquet (Apache columnar format)

```python
df = pd.read_parquet('data.parquet')
```

> Requires: `pyarrow` or `fastparquet`

---

## ✅ 5. SQL (Databases)

```python
import sqlite3

conn = sqlite3.connect('my_database.db')
df = pd.read_sql('SELECT * FROM my_table', conn)
```

> For other DBs like PostgreSQL, use `sqlalchemy` or `psycopg2`.

---

## ✅ 6. HTML (Web Tables)

```python
dfs = pd.read_html('https://example.com/page-with-tables')  # Returns list of DataFrames
```

> Requires: `lxml` or `html5lib`

---

## ✅ 7. Clipboard

```python
df = pd.read_clipboard()  # Paste tabular data copied from Excel or web
```

---

## ✅ 8. ORC (Optimized Row Columnar)

```python
df = pd.read_orc('data.orc')
```

> Requires: `pyarrow`

---

## ✅ 9. Feather

```python
df = pd.read_feather('data.feather')
```

> Fast binary format (good for pipeline stages)

---

## ✅ 10. Pickle (Python serialized object)

```python
df = pd.read_pickle('data.pkl')
```

---

## ✅ 11. XML

```python
df = pd.read_xml('data.xml')
```

> Requires: `lxml`

---

## ✅ 12. Google Sheets (via URL)

```python
sheet_url = 'https://docs.google.com/spreadsheets/d/xyz123/export?format=csv'
df = pd.read_csv(sheet_url)
```

> Ensure the sheet is **shared publicly or accessibly via link**

---

## ✅ 13. Reading from ZIP File (compressed)

```python
df = pd.read_csv('data.zip', compression='zip')
```

> Works similarly for gzip, bz2, xz, etc.

---

## ✅ 14. Reading from S3 / HTTPS / GCS / Azure Blob (Cloud URLs)

```python
df = pd.read_csv('https://bucket.s3.amazonaws.com/data.csv')
df = pd.read_parquet('gs://your-bucket/data.parquet')  # if gcsfs installed
```

> Requires: credentials + cloud FS library (e.g., `s3fs`, `gcsfs`, `adlfs`)

````python
# merge()

# The pd.merge() function implements, a number of types of joins:
# a. one to one join
# b. Many to one join
# c. Many to many join

# All three types of joins are assessed via an identical call to the pd.merge() Interface.
# The type of join performed depends on the form of input data.
````

````python
# One-to-One Join
# They are similar to column-wise concatenation.

df1 = pd.DataFrame(
    {"employee" : ["Bob", "Jake", "Lisa", "Sue"],
     "group" : ["Accounting", "Engineering", "Engineering", "HR"]}
)

df2 = pd.DataFrame(
    {"employee" : ["Lisa", "Bob", "Jake", "Sue"],
     "hire_date" : [2004, 2008, 2012, 2014]}
)
print(df1)
print("\n")
print(df2)
df3 = pd.merge(df1, df2)
df3
````

````python
# Many-to-One Join
# Many-to-One Joins Are the joints in which one of the two key columns contains duplicate entries.

# For the Many-to-One case, the resulting data frame will reserve those duplicate entries as appropriate.

df4 = pd.DataFrame(
    {"group" : ["Accounting", "Engineering", "HR"],
     "supervisor" : ["Carly", "Guido", "Steve"]}
)
print(df3)
print("\n")
print(df4)
pd.merge(df3, df4)
````

````python
# Many-to-Many Join
# Many-to-Many Joins Are a bit confusing conceptually, but are nevertheless, well defined.
# If the key column in both the left and the right array contains duplicates, then the result is a many to many merge.
df5 = pd.DataFrame(
    {"group" : ["Accounting", "Accounting", "Engineering", "Engineering", "HR", "HR"],
    "skills" : ["math", "spreadsheets", "coding", "linux", "spreadsheets", "organization"]}
)
print(df1)
print("\n")
print(df5)
pd.merge(df3, df5)
````

````python
print(df1)
print("\n")
print(df2)
pd.merge(df1, df2, on="employee")
# on - specify the column to join on
````

````python
df3 = pd.DataFrame(
    {"name": ["Bob", "Jake", "Lisa", "Sue"],
    "salary": [70000, 80000, 120000, 90000]}
)

print(df1)
print("\n")
print(df3)
# pd.merge(df1, df3, left_on="employee", right_on="name")
pd.merge(df1, df3, left_on="employee", right_on="name").drop("name", axis=1)
````

````python
df6 = pd.DataFrame(
    {"name": ["Peter", "Paul", "Mary"],
    "food": ["fish", "beans", "bread"]}
)
df7 = pd.DataFrame(
    {"name": ["Mary", "Joseph"],
    "drink": ["wine", "beer"]}
)
print(df6)
print("\n")
print(df7)
pd.merge(df6, df7) # by default, it will perform an inner join
````

````python
pd.merge(df6, df7, how="inner") # inner join
````

````python
# how - specify the type of join
# left - left join
# right - right join
# outer - outer join
# inner - inner join

print(df6)
print("\n")
print(df7)

pd.merge(df6, df7, how="outer") # outer join
````
