```python
# Hypothesis Testing
```

```python
# p-value: 
# p-value helps answer:
# “Is the observed result real, or could it have happened just by random chance?”

# In Simple terms:
# p-value measures:
# How surprising the observed result is if the null hypothesis were true.

# Example:
# Suppose:
# A student normally scores around:
# 50 marks

# Suddenly one day:
# Student scores 98.

# Teacher starts thinking:
# “Was this genuine improvement or just luck/chance?”

# That doubt is exactly what p-value tries to quantify.
```

```python
# p-value always works with:
# Null Hypothesis

# Null Hypothesis Means
# “Nothing special happened.”
# or
# “No effect exists.”

# Example — Medicine Trial
# Suppose company claims:
# New medicine reduces fever faster.

# Null Hypothesis
# Medicine has:
# No real effect.
# Any improvement happened by chance.

# p-value checks:
# “If medicine truly had NO effect, how likely is the observed result?”


# Small p-value
# Means:
# Observed result is VERY unlikely under null hypothesis.
# So we start doubting null hypothesis.

# Large p-value
# Means:
# Observed result is quite possible under null hypothesis.
# So no strong evidence exists against null hypothesis.

# Note: p-value is NOT probability that null hypothesis is true.
# Instead: p-value is:
# Probability of observing results this extreme (or more extreme), assuming null hypothesis is true.

# Real-Life Example — Coin Toss
# Suppose coin is assumed fair.

# Null Hypothesis
# Coin is fair.
# Probability of head:
# 50%

# Experiment
# You toss coin 10 times.
# Result:
# 10 heads out of 10.
# Now you think:
# “This seems suspicious.”

# p-value asks:
# If coin were truly fair,
# how likely is getting such extreme result?
```

```python
# Such outcome is extremely unlikely for fair coin.
# So we reject:
# Null hypothesis
```

### Typical Decision Rule

| p-value  | Interpretation               |
| -------- | ---------------------------- |
| p < 0.05 | Strong evidence against null |
| p > 0.05 | Weak evidence against null   |


### Real Industry Examples

| Industry             | Typical Threshold |
| -------------------- | ----------------- |
| Medical research     | 0.01 or stricter  |
| Business analytics   | 0.05              |
| Exploratory research | Sometimes 0.1     |


```python
# Real-Life A/B Testing Example
# Suppose Amazon tests:
# Old button
# New button

# New button increases conversion by 5%.

# Question:
# Is this improvement REAL?
# Or random fluctuation?

# p-value helps answer this.

# Small p-value
# Means:
# Improvement likely genuine.

# Large p-value
# Means:
# Difference may simply be random noise.

# p-value does NOT measure:
# - Importance
# - Business value
# - Effect size
```

```python
# Example
# Suppose:
# Huge dataset of 1 crore users.

# Tiny improvement:
# 0.1%

# may still produce:
# Very small p-value

# But business impact may be negligible.

# So Small p-value ≠ Big Impact Always
# VERY important real-world understanding.
```

```python
# Real-Life Healthcare Example
# Suppose new drug lowers blood pressure by:
# 0.5 mmHg

# p-value = 0.0001

# Statistically significant.

# But medically:
# Maybe not meaningful.
```

```python
from scipy.stats import ttest_1samp

data = [52,55,48,50,60]

t_stat, p_value = ttest_1samp(data, 50)
print(p_value)
```
