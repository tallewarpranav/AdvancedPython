```python
# Question:
# The average IQ for the adult population is 100 with a standard deviation of 15.
# A researcher believes this value has changed.
# The researcher decides to test the IQ of 75 random adults.
# The average IQ of the sample is 105.
# Is there enough evidence to suggest the average IQ has changed?

import math
from scipy.stats import norm

# Step 1: Define known values from the problem
mu = 100         # Population mean (claimed average IQ)
sigma = 15       # Population standard deviation
n = 75           # Sample size (number of adults tested)
x_bar = 105      # Sample mean (average IQ from the sample)

# Step 2: Compute the standard error of the mean (SE)
# Formula: SE = sigma / sqrt(n)
standard_error = sigma / math.sqrt(n)

# Step 3: Compute the z-score for the sample mean
# Formula: z = (x̄ - μ) / SE
z_score = (x_bar - mu) / standard_error

# Step 4: Compute the p-value for a two-tailed test
# Since it's a two-tailed test (μ ≠ 100), we calculate both sides of the tail
p_value = 2 * (1 - norm.cdf(abs(z_score)))

# Step 5: Decide whether to reject the null hypothesis using significance level alpha
alpha = 0.05  # Typical threshold for statistical significance
reject_null = p_value < alpha  # True if p-value is less than alpha

# Output results
print("Z-score:", z_score)
print("P-value:", p_value)
print("Reject Null Hypothesis?" , reject_null)
```

```python
# There is statistically significant evidence at the 5% level to suggest that the average IQ has changed from 100.
```

```python
# Another way to solve the above problem

import math
from scipy.stats import norm

# Step 1: Given values
mu = 100         # Population mean
sigma = 15       # Population standard deviation
n = 75           # Sample size
x_bar = 105      # Sample mean
alpha = 0.05     # Significance level for 2-tailed test

# Step 2: Calculate standard error (SE)
standard_error = sigma / math.sqrt(n)

# Step 3: Calculate test Z-score
z_score = (x_bar - mu) / standard_error

# Step 4: Get the critical Z-value for two-tailed test
# 1 - alpha/2 because it's a two-tailed test
z_critical = norm.ppf(1 - alpha / 2)

# Step 5: Compare test Z-score to critical Z-value
if abs(z_score) > z_critical:
    result = "Reject the null hypothesis: There is enough evidence to suggest the average IQ has changed."
else:
    result = "Fail to reject the null hypothesis: There is not enough evidence to suggest the average IQ has changed."

# Final outputs
print("Z-score:", round(z_score, 3))
print("Critical Z-value:", round(z_critical, 3))
print("Decision:", result)
```

```python
# Problem:
# The average IQ of the adult population is 100.
# A researcher believes the average IQ of adults is lower.
# A random sample of 5 adults is tested and scored:
# 69, 79, 89, 99, 109 (with sample standard deviation 15.81)
# Is there enough evidence to suggest the average IQ is lower?

import math
from scipy.stats import t

# Step 1: Define the known values
population_mean = 100         # μ: Claimed population mean
sample_mean = 89              # x̄: Sample mean from data
sample_std_dev = 15.81        # s: Sample standard deviation
sample_size = 5               # n: Sample size
alpha = 0.05                  # Significance level (left-tailed test) How? Because the researcher believes the average IQ is lower, we are conducting a left-tailed test.

# Step 2: Compute standard error of the mean
standard_error = sample_std_dev / math.sqrt(sample_size)  # s / sqrt(n)

# Step 3: Compute test statistic (t-score)
t_statistic = (sample_mean - population_mean) / standard_error  # (x̄ - μ) / (s / sqrt(n))

# Step 4: Compute the critical t-value for left-tailed test
# Degrees of freedom = n - 1
df = sample_size - 1
t_critical = t.ppf(alpha, df)  # For left tail

# Step 5: Compare test statistic to critical t-value
if t_statistic < t_critical:
    decision = "Reject the null hypothesis"
    interpretation = "There is enough statistical evidence to suggest the average IQ is lower than 100."
else:
    decision = "Fail to reject the null hypothesis"
    interpretation = "There is not enough statistical evidence to suggest the average IQ is lower than 100."

# Step 6: Print results
print("T-statistic:", round(t_statistic, 3))
print("Critical T-value:", round(t_critical, 3))
print("Decision:", decision)
print("Interpretation:", interpretation)
```

```python
# Problem:
# The NFL believes the chance of the Patriots winning the coin toss is greater than 50%.
# In a random sample of 200 tosses, the Patriots won 118 times.
# Is there enough evidence to suggest the Patriots are cheaters?

import math
from scipy.stats import norm

# Step 1: Define known values
sample_wins = 118            # Number of successes (Patriots win)
sample_size = 200            # Number of tosses (n)
sample_proportion = sample_wins / sample_size  # p̂ = x/n = 118/200
population_proportion = 0.50  # p₀: Assumed population proportion under H₀
alpha = 0.01                 # Significance level for 99% confidence
tail_type = "right-tailed"   # From problem statement

# Step 2: Calculate standard error for population proportion
standard_error = math.sqrt((population_proportion * (1 - population_proportion)) / sample_size)

# Step 3: Calculate Z statistic
z_statistic = (sample_proportion - population_proportion) / standard_error

# Step 4: Get critical z-value for right-tailed test
z_critical = norm.ppf(1 - alpha)  # one-tailed right-side

# Step 5: Compare and interpret
if z_statistic > z_critical:
    decision = "Reject the null hypothesis"
    interpretation = "There is enough evidence to suggest the Patriots are cheaters (p > 0.5)."
else:
    decision = "Fail to reject the null hypothesis"
    interpretation = "There is not enough evidence to prove the Patriots are cheaters."

# Step 6: Print all key results
print("Sample Proportion (p̂):", round(sample_proportion, 3))
print("Standard Error (SE):", round(standard_error, 4))
print("Z-statistic:", round(z_statistic, 3))
print("Critical Z-value (at α = 0.01):", round(z_critical, 3))
print("Decision:", decision)
print("Interpretation:", interpretation)
```

```python
# Problem: Chi-Square Test

# A casino claims that the way dice outcomes are distributed is fair.
# Total die throws = 204
# Observed frequencies:
# 1: 22, 2: 24, 3: 38, 4: 30, 5: 46, 6: 44

# Expected frequency for each face (assuming fair die) =
# 204 / 6 = 34


# We are to test:
# H₀ (Null Hypothesis): The die is fair.
# H₁ (Alternate Hypothesis): The die is not fair.

from scipy.stats import chi2

# Step 1: Observed and Expected frequencies
observed = [22, 24, 38, 30, 46, 44]  # From image
total_throws = 204
expected = [total_throws / 6] * 6  # Fair die → equal probabilities

# Step 2: Calculate Chi-square test statistic manually
chi_square_statistic = sum([(o - e)**2 / e for o, e in zip(observed, expected)])

# Step 3: Set significance level and degrees of freedom
alpha = 0.01
df = 6 - 1  # 6 faces → df = categories - 1

# Step 4: Critical value from chi-square distribution table
critical_value = chi2.ppf(1 - alpha, df)

# Step 5: Decision
if chi_square_statistic > critical_value:
    decision = "Reject the null hypothesis"
    interpretation = "There is enough evidence to suggest the die is not fair."
else:
    decision = "Fail to reject the null hypothesis"
    interpretation = "There is not enough evidence to suggest the die is unfair."

# Step 6: Print results
print("Chi-Square Statistic:", round(chi_square_statistic, 3))
print("Critical Value (α = 0.01, df = 5):", round(critical_value, 3))
print("Decision:", decision)
print("Interpretation:", interpretation)
```

```python
# Problem: Anova
# We want to measure the level of stress during layoffs.
# There are three groups:
# Group 1 (Normal): [2, 3, 7, 2, 6]
# Group 2 (Announced layoffs): [10, 8, 7, 5, 10]
# Group 3 (During layoffs): [10, 13, 14, 13, 15]
# Are the differences in stress levels statistically significant?

import numpy as np
import scipy.stats as stats

# Step 1: Define group data
group_normal = [2, 3, 7, 2, 6]
group_announced = [10, 8, 7, 5, 10]
group_during = [10, 13, 14, 13, 15]

# Combine all data
all_data = group_normal + group_announced + group_during

# Step 2: Calculate group means
mean_normal = np.mean(group_normal)
mean_announced = np.mean(group_announced)
mean_during = np.mean(group_during)
grand_mean = np.mean(all_data)

# Step 3: Calculate SSW (Sum of Squares Within Groups)
ssw = sum((x - mean_normal)**2 for x in group_normal) + \
      sum((x - mean_announced)**2 for x in group_announced) + \
      sum((x - mean_during)**2 for x in group_during)

# Step 4: Calculate SST (Total Sum of Squares)
sst = sum((x - grand_mean)**2 for x in all_data)

# Step 5: Calculate SSB (Between Groups) = SST - SSW
ssb = sst - ssw

# Step 6: Degrees of freedom
k = 3  # Number of groups
n = len(all_data)  # Total number of observations
df_between = k - 1
df_within = n - k

# Step 7: Mean Squares
ms_between = ssb / df_between
ms_within = ssw / df_within

# Step 8: F-statistic
f_value = ms_between / ms_within

# Step 9: p-value from F-distribution
p_value = 1 - stats.f.cdf(f_value, df_between, df_within)

# Step 10: Conclusion at alpha = 0.05
alpha = 0.05
if p_value < alpha:
    conclusion = "Reject the null hypothesis: Stress levels are significantly different across groups."
else:
    conclusion = "Fail to reject the null hypothesis: No significant difference in stress levels."

# Final output
print(f"Group Means: Normal={mean_normal}, Announced={mean_announced}, During={mean_during}")
print(f"Grand Mean: {grand_mean}")
print(f"SST: {round(sst, 1)}, SSW: {round(ssw, 1)}, SSB: {round(ssb, 1)}")
print(f"F({df_between}, {df_within}) = {round(f_value, 2)}, p = {round(p_value, 4)}")
print("Conclusion:", conclusion)
```

```python
# Ideal Code:
from scipy.stats import f_oneway

# Group data from the slide
group_normal = [2, 3, 7, 2, 6]
group_announced = [10, 8, 7, 5, 10]
group_during = [10, 13, 14, 13, 15]

# Apply one-way ANOVA (F-test)
f_statistic, p_value = f_oneway(group_normal, group_announced, group_during)

# Print results
print("F-statistic:", round(f_statistic, 2))
print("P-value:", round(p_value, 4))

# Interpret the results
alpha = 0.05
if p_value < alpha:
    print("Conclusion: Reject the null hypothesis — Stress levels differ significantly across groups.")
else:
    print("Conclusion: Fail to reject the null hypothesis — No significant difference in stress levels.")
```
