```python
# Matplotlib is a multi-platform data visualisation library rebuilt on the top of numpy array.
# One of the most important features of matplotlib is its ability to play well with many operating systems and graphical backends.
# This cross-platform, everything to everyone approach has been one of the greatest strengths of matplotlib.

# Matplotlib was originally developed by John D. Hunter in 2003.
# It was written as a Python alternative to MATLAB's plotting capabilities, and much of its syntax and functionality is similar to MATLAB.

# In the recent years, matplotlib has  begun to show its age. Newer tools like Plotly, ggplot, and seaborn have emerged that 
# offer more modern features and better aesthetics, making matplotlib feel clunky and old-fashioned.
# Still, I am of the opinion that we cannot ignore matplotlib's strength as a well-tested, cross-platform graphics engine.
```

```python
fig, ax = plt.subplots(2, 2) # create a figure and a set of subplots
# fig: figure object
# ax: array of axes objects
# ax[0, 0]: first subplot
# ax[0, 1]: second subplot
# ax[1, 0]: third subplot
# ax[1, 1]: fourth subplot

ax[0, 0].plot(x, y) # plot on the first subplot
ax[0, 1].plot(x, np.cos(x)) # plot on the second subplot
ax[1, 0].plot(x, np.tan(x)) # plot on the third subplot
ax[1, 1].plot(x, np.tanh(x)) # plot on the fourth subplot
plt.show()
```

```python
# Lets get started with Data Visualization using Matplotlib
```

```python
tips.groupby('day')['total_bill'].mean()
```

```python
tips.groupby('day')['total_bill'].sum()
```

```python
# Bar Plot (bivariate plot)
# Use Case:
# Bar plots are ideal for comparing quantities across categories and they are very common in exploratory data analysis.
# Syntax:
# plt.bar(x, height, width, color)

# Total bill sum by day
bill_by_day = tips.groupby('day')['total_bill'].sum()
# print(bill_by_day)
# print(bill_by_day.index) # Index(['Thur', 'Fri', 'Sat', 'Sun'], dtype='object', name='day')
# print(bill_by_day.values) # Values([  1096.33, 325.88, 1778.40, 1627.16])

plt.figure(figsize=(6,4))
plt.bar(bill_by_day.index, bill_by_day.values)
plt.title("Total Bill by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill Amount")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(bill_by_day.index, bill_by_day.values, color='orange', width=0.5)
plt.title("Total Bill by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill Amount")
plt.show()
```

```python
# Insight: Most of the total bill is on Saturday and Sunday
```
