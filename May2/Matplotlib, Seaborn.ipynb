````python
# Bar Plot (bivariate plot)
# Use Case:
# Bar plots are ideal for comparing quantities across categories and they are very common in exploratory data analysis.
# Syntax:
# plt.bar(x, height, width, color)

# Total bill sum by day
bill_by_day = tips.groupby('day')['total_bill'].sum()
# print(bill_by_day)
# print(bill_by_day.index) # Index(['Thur', 'Fri', 'Sat', 'Sun'], dtype='object', name='day')
# print(bill_by_day.values) # Values([  1096.33, 325.88, 1778.40, 1627.16])

plt.figure(figsize=(6,4))
plt.bar(bill_by_day.index, bill_by_day.values)
plt.title("Total Bill by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill Amount")
plt.show()
````

````python
plt.figure(figsize=(6,4))
plt.bar(bill_by_day.index, bill_by_day.values, color='orange', width=0.5)
plt.title("Total Bill by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill Amount")
plt.show()
````

````python
# Insight: Most of the total bill is on Saturday and Sunday
````

````python
# Histogram (univariate plot)
# Use Case:
# Histograms are used to visualize the distribution of a single numerical variable.
# It is essential for understanding the underlying frequency distribution of the data, skewness, or modality.
# Syntax:
# plt.hist(x, bins=10, color='color_name', alpha=0.5, edgecolor='black')

# Total bill histogram
plt.figure(figsize=(6,4))
plt.hist(tips['total_bill'], color='orange', alpha=0.9, edgecolor='black')
# plt.hist(tips['total_bill'], color='orange', alpha=0.9, edgecolor='pink')
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill Amount")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
````

````python
# Scatter Plot (bivariate plot)
# Use Case:
# Scatter plots are used to visualize the relationship between two numerical variables.
# They are useful for identifying correlations, trends, and regression analysis.
# Syntax:
# plt.scatter(x, y, color='purple', alpha=0.5, s=100, marker='o')
# s: size of the marker
# marker: shape of the marker
# alpha: transparency of the marker
# color: color of the marker

# Total bill vs tip scatter plot
plt.figure(figsize=(6,4))
plt.scatter(tips['total_bill'], tips['tip'], color='orange', alpha=0.8)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill Amount")
plt.ylabel("Tip Amount")
plt.grid()
plt.show()
````

````python
# Pie Chart (univariate plot)
# Use Case:
# Pie charts represent parts of a whole. They are used for category proportions visualizations.
# Less used in data science, due to their limitations in accurately representing data.
# Syntax:
# plt.pie(values, labels=labels, colors, autopct='%1.1f%%', startangle=140, explode=explode)
# values: data to be plotted
# labels: labels for the data
# colors: colors for the data
# autopct: format for the percentage
# startangle: angle to start the pie chart
# explode: to separate the slices

# Percentage of smokers and non-smokers
smoker_counts = tips['smoker'].value_counts()
print(smoker_counts)
# print(tips['smoker'].value_counts(normalize=True)*100)

plt.figure(figsize=(6,4))
plt.pie(smoker_counts)
plt.title("Percentage of Smokers and Non-Smokers")
plt.show()
````

````python
# Horizontal Bar Plot (bivariate plot)
# Use Case:
# Horizontal bar plots are used when category names are long or when you want to emphasize relative lengths horizontally.
# Syntax:
# plt.barh(y, width, color='color_name', alpha=0.5, edgecolor='black')
# y: y-axis values
# width: width of the bars
# color: color of the bars
# alpha: transparency of the bars
# edgecolor: color of the edges of the bars

# Average tip by day
avg_tip_by_day = tips.groupby('day')['tip'].mean()
print(avg_tip_by_day)

plt.figure(figsize=(6,4))
plt.barh(avg_tip_by_day.index, avg_tip_by_day.values, color='orange', alpha=0.9, edgecolor='black')
plt.title("Average Tip by Day")
plt.xlabel("Average Tip Amount")
plt.ylabel("Day of the Week")
plt.grid(axis='x', linestyle='--', alpha=0.3)
plt.show()
````

````python
# Insight from the horizontal bar plot:
# The average tip is highest on Sunday and lowest on Thursday.
````

````python
# Line Plot (bivariate plot)
# Use Case:
# Line plots are used to visualize trends or changes over a continuous variable, typically time or index.
# They are often used for tracking model training metrics or time series data.
# Syntax:
# plt.plot(x, y, color='color_name', label='label_name', linestyle='-', linewidth=2, marker='o')
# x: x-axis values
# y: y-axis values
# color: color of the line
# label: label for the line
# linestyle: style of the line
# linewidth: width of the line
# marker: shape of the marker
# x: x-axis values
# y: y-axis values
# color: color of the line
# label: label for the line
# linestyle: style of the line
# linewidth: width of the line
# marker: shape of the marker

# Average tip amount over table size
avg_tip_by_size = tips.groupby('size')['tip'].mean()
print(avg_tip_by_size)

plt.figure(figsize=(6,4))
plt.plot(avg_tip_by_size.index, avg_tip_by_size.values, color='orange', label='Average Tip', linestyle='-', linewidth=2, marker='o')
plt.title("Average Tip by Table Size")
plt.xlabel("Table Size")
plt.ylabel("Average Tip Amount")
plt.grid()
plt.legend() # Add legend to the plot
plt.show()
````

````python
# Insight: The average tip amount increases with the table size.
# The average tip amount is highest for table size 6 and lowest for table size 1.
````

````python
# Box Plot (univariate plot)
# Use Case:
# Box plots are used to visualize the distribution of a numerical variable and identify outliers.
# They are useful for comparing distributions across categories. They are frequently used in feature analysis for machine learning.
# Syntax:
# plt.boxplot(data, vert=True, patch_artist=True, notch=False, widths=0.5)
# data: data to be plotted
# vert: vertical or horizontal box plot
# patch_artist: fill the box with color. eg: patch_artist=True
# notch: whether to show the notch in the box plot. eg: notch=True
# widths: width of the box. eg: widths=0.5

# Tip amount box plot
plt.figure(figsize=(6,4))
plt.boxplot(tips['tip'])
plt.title("Box Plot of Tip Amount")
plt.xlabel("Tip Amount")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
````

````python

### **Step 1: Code to Load the `penguins` Dataset**
#import seaborn as sns

## Load the penguins dataset
penguins = sns.load_dataset("penguins")

# Show first few rows
print(penguins.head())


### **Step 2: Explanation of Columns in `penguins` Dataset**

#| Column Name         | Description                                        |
#| ------------------- | -------------------------------------------------- |
#| `species`           | Type of penguin (e.g., Adelie, Chinstrap, Gentoo)  |
#| `island`            | Island where the penguin was located               |
#| `bill_length_mm`    | Length of the penguin’s bill (beak) in millimeters |
#| `bill_depth_mm`     | Depth (thickness) of the bill in millimeters       |
#| `flipper_length_mm` | Length of the flipper in millimeters               |
#| `body_mass_g`       | Weight of the penguin in grams                     |
#| `sex`               | Gender of the penguin (Male or Female)             |
#| `year`              | Year when the data was recorded                    |

penguins.describe()

### **Step 3: Tasks**

#### **Task 1: Line Plot**

#**Instruction:** Plot the **average flipper length for each species** using a line chart to see how species differ in flipper size.
#**Hint:** Use `plt.plot()` and group by `species`.
plt.figure(figsize=(6,4))
avg_flipper_length = penguins.groupby('species')['flipper_length_mm'].mean()
plt.plot(avg_flipper_length.index, avg_flipper_length.values, marker='o', linestyle
='-', color='orange')
plt.title("Average Flipper Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Flipper Length (mm)")
plt.grid()
plt.show()

#### **Task 2: Scatter Plot**
#**Instruction:** Create a scatter plot to examine if **bill length and bill depth** are correlated.
#**Hint:** Use `plt.scatter()` with `x = bill_length_mm` and `y = bill_depth_mm`.
print("Create a scatter plot to examine if bill length and bill depth are correlated.")
plt.figure(figsize=(6,4))
plt.scatter(penguins['bill_length_mm'], penguins['bill_depth_mm'], color='green')
plt.title("Bill Length vs Bill Depth")
plt.xlabel("Bill Length (mm)")
plt.ylabel("Bill Depth (mm)")
plt.grid()
plt.show()
print("Correlation between bill length and bill depth:", penguins['bill_length_mm'].corr(penguins['bill_depth_mm']))

#### **Task 3: Bar Chart**
#**Instruction:** Create a bar chart showing the **average body mass for each gender**.
#**Hint:** Use `plt.bar()` with grouped values by `sex` (Male/Female).
print("Create a bar chart showing the average body mass for each gender.")
avg_body_mass = penguins.groupby('sex')['body_mass_g'].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_body_mass.index, avg_body_mass.values, color=['blue', 'pink'])
plt.title("Average Body Mass by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Body Mass (g)")
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()

#### **Task 4: Pie Chart**

#**Instruction:** Show the **percentage of penguins found on each island** using a pie chart.
#**Hint:** Use `plt.pie()` with the `island` value counts.
print("Show the percentage of penguins found on each island using a pie chart.")
island_counts = penguins['island'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(island_counts, labels=island_counts.index, autopct='%1.1f%%', colors=['orange', 'blue', 'green'])
plt.title("Percentage of Penguins by Island")
plt.show()

#### **Task 5: Histogram**

#**Instruction:** Plot a histogram to see the **distribution of flipper lengths** across all penguins.
#**Hint:** Use `plt.hist()` with `flipper_length_mm`.
print("Plot a histogram to see the distribution of flipper lengths across all penguins.")
plt.figure(figsize=(6,4))
plt.hist(penguins['flipper_length_mm'], bins=15, color='purple', alpha=0.7, edgecolor='black')
plt.title("Distribution of Flipper Lengths")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
````
