```python
# Parametric vs Non-Parametric Hypothesis Tests

# Suppose you want to test:
# - Is new medicine better?
# - Is new website improving sales?
# - Are two machine outputs different?
# - Did training improve employee performance?

# We use: Hypothesis Testing!

# But now comes a BIG question:
# “What type of test should we use?”

# That depends on:
# Nature of data and assumptions.

# This creates two categories:
```

| Type                 | Meaning                           |
| -------------------- | --------------------------------- |
| Parametric Tests     | Assume specific data distribution |
| Non-Parametric Tests | Make fewer/no strict assumptions  |


```python
# Parametric Tests
# = “I trust the data follows certain mathematical rules.”

# Non-Parametric Tests
# = “I don’t trust the distribution assumptions.”
```

```python
# What Are Parametric Tests?

# Parametric tests assume:
# - Data follows normal distribution
# - Mean and standard deviation are meaningful
# - Data is continuous
# - Variance behaves properly

# Parametric tests work well when:
# Data is clean, structured, and approximately normal.

# Real-Life Example — Manufacturing
# Suppose machine bolt lengths are:
#     9.9 mm
#     10.1 mm
#     10.0 mm

# This data is:
# - Continuous
# - Stable
# - Near normal distribution

# Perfect for: Parametric tests
```

### Common Parametric Tests

| Test                | Purpose                 |
| ------------------- | ----------------------- |
| t-test              | Compare means           |
| ANOVA               | Compare multiple groups |
| Pearson correlation | Relationship analysis   |
| Z-test              | Mean comparison         |


```python
# Example — Employee Training

# Suppose:
# You measure employee productivity before and after training.
# Data looks normally distributed.

# You use:
# Paired t-test (parametric test)
```

```python
# But There Is a Problem

# Real-world business data is often:
# - Skewed
# - Messy
# - Contains outliers
# - Non-normal

# Then parametric assumptions break.

# Example — Salary Data
# Salaries are highly skewed.

# Few CEOs earn huge amounts.

# Mean gets distorted.

# Normality assumption fails.

# Using parametric tests becomes risky.

# Solution?
# Non-Parametric Tests
```

```python
# What Are Non-Parametric Tests?

# These tests:
# Do NOT strongly assume normal distribution.
# They are more flexible.

# Non-parametric tests work when:
# - Data is skewed
# - Outliers exist
# - Data is ordinal/categorical
# - Sample size is small
# - Distribution unknown

# Real-Life Example — Customer Ratings
# Ratings:
#     Poor
#     Average
#     Good
#     Excellent

# This is: Ordinal data
# Mean is not very meaningful.
# So we prefer: Non-parametric tests
```

### Common Non-Parametric Tests

| Test                 | Purpose               |
| -------------------- | --------------------- |
| Mann-Whitney U Test  | Alternative to t-test |
| Wilcoxon Test        | Paired comparison     |
| Kruskal-Wallis Test  | Alternative to ANOVA  |
| Spearman Correlation | Rank relationship     |


```python
# Real-Life Example — Swiggy Ratings
# Suppose customers rate delivery experience: 1–5 stars.

# This data is:
# - Ordinal
# - Skewed
# - Non-normal

# Better suited for: Non-parametric testing
```

| Parametric                        | Non-Parametric                 |
| --------------------------------- | ------------------------------ |
| Assumes normality                 | No strict normality assumption |
| Uses mean & SD                    | Often uses ranks/medians       |
| More powerful if assumptions true | More robust                    |
| Sensitive to outliers             | Less sensitive                 |


### Use Parametric Tests When:

| Scenario                   | Reason                 |
| -------------------------- | ---------------------- |
| Manufacturing measurements | Normally distributed   |
| Sensor readings            | Continuous stable data |
| Controlled experiments     | Assumptions hold       |


### Use Non-Parametric Tests When:

| Scenario                | Reason                 |
| ----------------------- | ---------------------- |
| Salary analysis         | Skewed                 |
| Customer ratings        | Ordinal                |
| Medical severity scales | Non-normal             |
| Small samples           | Distribution uncertain |


# Non-parametric Statistical hypothesis tests that you need in applied machine learning

## Wilcoxon Sign Test:

It is used to compare the two paired Samples

Paired Samples: Same people, two different situations

H0: There is no significant difference in the calcium level of the patient initial and after 1 weeks.

H1: There is significant difference in the calcium level of the patient initial and after 1 weeks.

If p < 0.05 -> Reject H0

If p >= 0.05 -> Accept H0

```python
import pandas as pd
df = pd.read_excel("Wilcoxon.xlsx", sheet_name=0)

df.head()
```

```python
from scipy.stats import wilcoxon

stats, p = wilcoxon(df.Before_Treatment, df.After_1_Week_of_Treatment)

print("Stats= ",stats,"\np value= ",p)
```

Since,  p >= 0.05 -> Accept H0

H0: There is no si
gnificant difference in the calcium level of the patient initial and after 1 weeks.

```python
from scipy.stats import wilcoxon

stats, p = wilcoxon(df.After_1_Week_of_Treatment, df.After_2_Week_of_Treatment)

print("Stats= ",stats,"\np value= ",p)
```

Since p < 0.05 -> Reject H0

H0: There is no significant difference in the calcium level of the patient after 1 week and after 2 weeks.

## Friedman Test

It is used to compare more than 2 paired samples.

H0: There is no significant difference in the calcium level of the patient initial and after 1 week and also after 2 weeks.

H1: There is significant difference in the calcium level of the patient initial and after 1 week and also after 2 weeks.

If p < 0.05 -> Reject H0

If p >= 0.05 -> Accept H0

```python
from scipy.stats import friedmanchisquare

stats, p = friedmanchisquare(df.Before_Treatment, df.After_1_Week_of_Treatment, df.After_2_Week_of_Treatment)

print("Stats= ", stats, "\np value= ", p)
```

Since p < 0.05 -> Reject H0

H0: There is no significant difference in the calcium level of the patient initial and after 1 week and also after 2 weeks.

## Mann Whitneyu Test

It is used to compare two independent samples.

H0: There is no significant difference in the sales of Design1 and Design2.

H1: There is significant difference in the sales of Design1 and Design2.

If p < 0.05 -> Reject H0

If p >= 0.05 -> Accept H0

```python
import pandas as pd
df = pd.read_excel("Mann_Whitney.xlsx", sheet_name=1)

df.head()
```

```python
from scipy.stats import mannwhitneyu

stats, p = mannwhitneyu(df.Design1, df.Design2)

print("Stats= ",stats,"\np value= ",p)
```

Since, p >= 0.05 -> Accept H0

H0: There is no significant difference in the sales of Design1 and Design2.

## Kruskal Wallis Test:

It is used to compare the two or more independent Samples

H0: There is no significant difference in the sales of Design1, Design2 and Design3.

H1: There is significant difference in the sales of Design1, Design2 and Design3.

If p < 0.05 -> Reject H0

If p >= 0.05 -> Accept H0

```python
import pandas as pd
df = pd.read_excel("Mann_Whitney.xlsx", sheet_name=2)

df.head()
```

```python
from scipy.stats import kruskal

stats, p = kruskal(df.Design1, df.Design2, df.Design3)

print("Stats= ",stats,"\np value= ",p)
```

Since, p >= 0.05 -> Accept H0

H0: There is no significant difference in the sales of Design1, Design2 and Design3.

## Chi Square Test:

It is used to check the dependency of the variables. Variables should be Categorical.

Variables:

Gender: Male(1) and Female(0)

Smoking: Past Smoker(2), Current Smoker(1) and Non Smoker(0)

H0: There is no dependency between gender and smoking.

H1: There is dependency between gender and smoking.

If p < 0.05 -> Reject H0

If p >= 0.05 -> Accept H0

```python
import pandas as pd
df = pd.read_excel("Chi_Square_Test.xlsx", sheet_name=0)

df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()

df.isnull().sum()
```

```python
df.head()
```

```python
from scipy.stats import chi2_contingency

tab = pd.crosstab(df.Gender, df.Smoking)

tab
```

```python
stats, p, dof, expected = chi2_contingency(tab)

print("Stats= ",stats,"\np value= ",p)
```

Since, p >= 0.05 -> Accept H0

H0: There is no dependency between gender and smoking.

## One Sample T-Test

It is used to compare sample mean with population mean.

Sample Mean ->

Population Mean: Lets assume this as 65

H0: There is no significant difference of mean of students heights against population mean = 65.

H1: There is significant difference of mean of students heights against population mean = 65.

```python
import pandas as pd
df = pd.read_excel("One Sample.xlsx", sheet_name=0)

df.head()
```

```python
from scipy.stats import ttest_1samp

stats, p = ttest_1samp(df.Height, 65)

print("Stats= ",stats,"\np value= ",p)
```

Since p < 0.05 -> Reject H0

## One way Anova Test

H0:  because of Promotion, there is a no significant impact on the Sales.

H1:  because of Promotion, there is a significant impact on the Sales

```python
import pandas as pd
df = pd.read_excel("ANCOVA1.xlsx", sheet_name=0)

df.head()
```

```python
df.columns

# Index(['Store Number', 'Sales', 'Promotion', 'Coupon', 'ClietelRatings'], dtype='object')
# Store Number: Unique identifier for each store.
# Sales: Total sales amount for each store.
# Promotion: Binary variable indicating if the store had a promotion (1) or not (0).
# Coupon: Binary variable indicating if the store offered coupons (1) or not (0).
# ClietelRatings: Customer ratings for the store (1 to 10 scale).
```

Sales -> Dependent Variable

Promotion -> Independent Variable

```python
df.Promotion.value_counts()
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols # OLS is Ordinary Least Square
model = ols('Sales~C(Promotion)',df).fit() # C means Promotion is Categorical variable
```

```python
oneway = sm.stats.anova_lm(model, type=2)
# type=1 is Regression
# type=2 is for ANOVA and ANCOVA
# type=3 is for Discriminant Analysis

print(oneway)
```

Between Variation is 106.066667 and within variation is 79.800000  

p-value is 0.000011 which is <0.05

Therefore we can say because of Promotion, there is a significant impact on the Sales.

## Two way Anova Test

```python
model = ols('Sales~C(Promotion)+C(Coupon)',df).fit()
```

```python
twoway = sm.stats.anova_lm(model, type=2)

print(twoway)
```

p value of Promotion is 8.032739e-10 and Coupon is 1.095036e-07 i.e. both of these are less than 0.05

Therefore we can say that both the variables are significantly affecting the Sales.

## Ancova

Dependent Variable -> Sales

Independent Variable -> Promotion, Coupon, Client Rating

```python
model = ols('Sales~C(Promotion)+C(Coupon)+ClietelRatings',df).fit()
```

```python
ancova = sm.stats.anova_lm(model, type=2)

print(ancova)
```

p value of Promotion is 1.301539e-09 and Coupon is 1.471502e-07 i.e. both of these are less than 0.05

whereas p value of ClietelRatings is 3.745789e-01 which is >0.05

Therefore we can say that both Promotion and Coupon variables are significantly affecting the Sales whereas ClietelRatings have no significant impact on the Sales.

```python
# Read the CSV and Display the Data
import pandas as pd

# Load dataset
df = pd.read_csv("b2b_model_test_data.csv")

df.shape  # Check the shape of the DataFrame
```

```python
df
#  B2B scenario
# -Two ML models (A & B) predict lead conversion.
# -Each client has a Region and an actual conversion status (Lead_Converted).
```

```python
# Paired t-Test: Is Model A Better Than Model B?
# Objective
# We want to compare Model A and Model B’s performance using the same clients.

# Hypothesis
# H₀: Mean accuracy of Model A = Model B
# H₁: Mean accuracy is different (two-tailed)

# Importing necessary libraries
from scipy.stats import ttest_rel
# ttest_rel - Paired t-test for related samples
# This test compares the means of two related groups to determine if there is a statistically significant difference between them.

# Paired t-test
t_stat, p_value = ttest_rel(df["Model_A_Accuracy"], df["Model_B_Accuracy"])

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
# If p < 0.05, we reject H₀.
if p_value < 0.05:
    print("Reject H₀: There is a significant difference between Model A and Model B.")
else:
    print("Fail to reject H₀: No significant difference between Model A and Model B.")
```

```python
# If rejected: There's strong evidence that Model A and Model B differ significantly.
```

```python
# ML Practical Importance
# Model Validation: Confirms if switching models leads to statistically better performance.
# Benchmarking: Avoid decisions based on chance from random test splits.
```

```python
df.head()  # Display the first few rows of the DataFrame
```

```python
# Chi-Square Test: Is Region Related to Conversion?
# Objective
# Determine whether Region has a significant impact on Lead_Converted.

# Hypothesis
# H₀: Region and conversion are independent
# H₁: Region and conversion are dependent

# Importing necessary libraries
from scipy.stats import chi2_contingency

# Create a contingency table
contingency_table = pd.crosstab(df["Region"], df["Lead_Converted"])

contingency_table
```

```python
# Run Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-square value:", chi2)
print("p-value:", p)
```

```python
# If p < 0.05, region influences conversion rate.
# Otherwise, region has no impact.
if p < 0.05:
    print("Reject H₀: Region and conversion are dependent.")
else:
    print("Fail to reject H₀: Region and conversion are independent.")
```

```python
# ML Practical Importance
# Feature Selection: Remove unimportant categorical variables.
# Customer Segmentation: Identify regions with better conversion for targeted marketing.
```

```python
# Shapiro-Wilk Test: Is Accuracy Normally Distributed?
# Objective
# Check if Model A’s accuracy is normally distributed to validate assumptions for t-test.

# shapiro-wilk test is a statistical test that checks if a sample comes from a normally distributed population.
from scipy.stats import shapiro

# Test for normality
stat, p = shapiro(df["Model_A_Accuracy"])

# Printing the results of the Shapiro-Wilk test
print("Shapiro-Wilk stat:", stat)
print("p-value:", p)
```

```python
# If p > 0.05: Data is normal → You can use parametric tests like t-test
# If p < 0.05: Data is not normal → Use non-parametric tests like Mann-Whitney U test
if p > 0.05:
    print("Data is normally distributed.")
else:
    print("Data is not normally distributed. Consider using non-parametric tests.")
# ML Practical Importance
# Ensures Validity of statistical tests.
# Improves Interpretability when plotting model accuracy distributions.
```

```python
# Mann–Whitney U Test (if data is not normal)
# Objective
# Compare two model accuracies without assuming normal distribution.

from scipy.stats import mannwhitneyu

u_stat, p_val = mannwhitneyu(df["Model_A_Accuracy"], df["Model_B_Accuracy"])
print("Mann–Whitney U stat:", u_stat)
print("p-value:", p_val)

# Same as t-test, but robust to outliers and non-normality.

# ML Practical Importance
# Safer when working with small datasets or skewed performance metrics.
```

```python
# Kolmogorov-Smirnov (KS) Test
# Objective
# Compare distribution of model predictions to detect shift or drift.
# (We use accuracy here as proxy since we don’t have probability predictions.)

from scipy.stats import ks_2samp

ks_stat, ks_pval = ks_2samp(df["Model_A_Accuracy"], df["Model_B_Accuracy"])
print("KS statistic:", ks_stat)
print("p-value:", ks_pval)
```

```python
# If p < 0.05: Distributions differ → possible drift or performance shift.
# If p > 0.05: Models behave similarly.
if ks_pval < 0.05:
    print("Reject H₀: Model A and Model B have different distributions.")
else:
    print("Fail to reject H₀: Model A and Model B have similar distributions.")

# ML Practical Importance
# Used for data drift detection, especially in real-time ML pipelines.
```

| Test Name      | Used For                         | ML Relevance                            |
| -------------- | -------------------------------- | --------------------------------------- |
| t-Test         | Compare model means              | A/B testing, performance validation     |
| Chi-Square     | Test feature-target independence | Feature selection                       |
| Shapiro-Wilk   | Normality test                   | Validates use of parametric tests       |
| Mann-Whitney U | Non-parametric model comparison  | Robust model evaluation                 |
| KS Test        | Distribution comparison          | Drift detection, model shift monitoring |
