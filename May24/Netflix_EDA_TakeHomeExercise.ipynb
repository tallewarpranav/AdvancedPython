## Column Description

The dataset contains the following columns:

| Column Name | Description |
|---|---|
| `show_id` | Unique identifier assigned to each title in the dataset. |
| `type` | Indicates whether the content is a `Movie` or a `TV Show`. |
| `title` | Name of the movie or TV show. |
| `director` | Director or directors associated with the title. This field can contain missing values, especially for TV shows. |
| `cast` | Main cast members listed for the title. Multiple names are usually separated by commas. |
| `country` | Country or countries associated with the production or availability of the title. Some titles may be linked to multiple countries. |
| `date_added` | Date on which the title was added to Netflix. This is useful for studying Netflix acquisition and catalog expansion patterns. |
| `release_year` | Year in which the movie or TV show was originally released. |
| `rating` | Maturity or content rating, such as `TV-MA`, `PG-13`, `TV-14`, or similar labels. |
| `duration` | Runtime for movies in minutes, or number of seasons for TV shows. |
| `listed_in` | Genre or category labels assigned to the title. A single title can belong to multiple genres. |
| `description` | Short textual summary of the title. This can be useful for text-based analysis, keyword extraction, or content positioning. |

Important note: Some columns, such as `cast`, `country`, and `listed_in`, contain multiple values inside a single cell. These columns may need cleaning or transformation before deeper analysis.
