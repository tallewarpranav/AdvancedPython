```python
# Common Terms Used in Statistics

# 1. Population and Sample:
# Population: The complete collection of data or all possible observations i.e. Everything you are interested in studying.
# eg: Suppose Siemens has:
# 50,000 employees worldwide
# If HR wants to study employee satisfaction:
# All 50,000 employees = Population
```

```python
# Problem with Population Analysis

# In real life:
# Analyzing entire population is often:
# - Expensive
# - Time-consuming
# - Difficult
# So companies use samples.
```

```python
# Sample: A small subset selected from the population.
# eg: 
# Out of 50,000 employees:
# Company surveys only 2,000 employees.
# Those 2,000 employees = Sample

# Advantages of Sampling:
# It saves:
# - Time
# - Cost
# - Effort
# - Resources
```

```python
# Example of Bad Sampling

# Suppose:
# You want to know average salary in India.
# But you survey only Mumbai IT professionals.
# Will result represent India properly?
# No.
# Because sample is biased.
```

```python
# Sampling is heavily used in:

# Machine Learning
# A/B Testing
# Surveys
# Clinical trials
# Fraud detection
```

```python
# Measures of Central Tendency
# These measures help us identify the “center” or “typical value” of the data.

# Suppose a manager asks:
# “On average, how many products are sold daily?”
# or
# “What is the typical salary in the company?”
# or
# “What is the usual delivery time?”

# We need one value that summarizes the dataset.
# That central representative value is called: Central Tendency

# Suppose employee salaries are:
# 40k, 42k, 43k, 45k, 44k

# Most values are around 43k–44k.
# So center is around that region.
```

```python
# Mean
# Also called: Average

# The mean is calculated by:
# Adding all values and dividing by total number of values.

# Suppose 5 students scored:
# 70, 80, 90, 60, 100

# To find average marks:

# Step 1 → Add all marks
# 70 + 80 + 90 + 60 + 100 = 400

# Step 2 → Divide by number of students
# 400 ÷ 5 = 80

# So:
# Mean = 80
```

```python
# Mean can work with:

# Discrete Data
# Examples:
# - Number of defects
# - Number of orders

# Continuous Data
# Examples:
# - Height
# - Weight
# - Temperature
```

```python
# Mean is sensitive to outliers and skewed data.

# What Is an Outlier?
# An outlier is:
# An unusually large or small value.

# Example:
# Suppose salaries are:
# 40k, 42k, 43k, 44k, 10 crore

# Now average becomes extremely high.

# But most employees still earn around 40k.

# So mean becomes misleading.

# Real life example:
# Suppose company CEO salary is extremely high.
# Average company salary may look huge.
# But it does NOT represent normal employees.
```

```python
# Why This Matters in Data Science?

# Outliers can badly affect:
# - ML models
# - Business reports
# - Forecasting
# - Decision-making

# So sometimes:
# Median is preferred over mean.
```

```python
import statistics
statistics.mean([4,89,54,-7,-9,27,5])
```

```python
# Important Limitation of Mean

# Mean works best when:
# - Data is balanced
# - No extreme outliers exist

# Mean performs poorly when:
# - Data is highly skewed
# - Extreme values exist
```

```python
# What Is Median?

# Median means:
# The middle value of a dataset after arranging the data in sorted order.

# Median tells us:
# “What value lies exactly at the center?”

# Median is VERY useful when:
# - Outliers exist
# - Data is skewed
# - Extreme values distort averages

# Example:
# Suppose salaries are:
# 40k, 42k, 43k, 44k, 10 crore

# Mean Problem
# Average becomes extremely high because of CEO salary.
# But does that represent most employees?
# No.

# Median Solution

# Sorted salaries:
# 40k, 42k, 43k, 44k, 10 crore

# Middle value = 43k
# This better represents normal employees.

# Median is:
# Less sensitive to outliers.
# That is why it is heavily used in real business analytics.
```

| Case                  | Method                       |
| --------------------- | ---------------------------- |
| Odd number of values  | Exact middle value           |
| Even number of values | Average of two middle values |


```python
# CASE 1 — Odd Number of Values
# Dataset:
# 5, 76, 98, 32, 1, -6, 34, 3, -65

# Step 1 — Sort the Data
# Ascending order:
# -65, -6, 1, 3, 5, 32, 34, 76, 98

# Step 2 — Find Middle Value
# Total numbers = 9

# Middle position:
# (9 + 1)/2 = 5th position

# Step 3 — Identify 5th Value
# 5th value = 5

# So:
# Median = 5
```

```python
# CASE 2 — Even Number of Values
# Dataset:
# 5, 76, 98, 32, 1, -6, 34, 3, -65, 99

# Step 1 — Sort Data
# -65, -6, 1, 3, 5, 32, 34, 76, 98, 99

# Step 2 — Total Values
# Total = 10
# Even number of observations.
# So:
# No single middle value exists.

# Step 3 — Find Two Middle Values
# Middle values are:
# 5 and 32

# Step 4 — Take Average
# Formula:
# Median=
# (5 + 32) / 2

# Result:
# 18.5

# Final Answer
# Median = 18.5
```

```python
# What Is Mode?

# Mode means:
# The value that occurs most frequently in a dataset.

# Mode tells us:
# “Which value appears the most?”

# Real-Life Understanding
# Suppose you ask:
# “Which mobile brand is most preferred?”
# or
# “Which pizza flavor sells most?”
# or
# “Which defect occurs most frequently in factory machines?”

# The answer comes from:
# Mode
```

```python
# Example — Swiggy
# Restaurant orders:
# Pizza, Burger, Pizza, Pizza, Sandwich

# Most frequent item:
# Pizza

# So:
# Mode = Pizza

# This helps:
# Inventory planning
# Marketing
# Recommendation systems
```

```python
# Note1:
# Mode can sometimes be misleading.

# Because:
# Most frequent value may NOT represent actual center.

# Example
# Dataset:
# 1, 1, 1, 50, 60, 70, 80

# Mode = 1

# But does 1 represent center of data?
# No.

# Most data lies much higher.
# So mode may sometimes give distorted understanding.
```

```python
# Note2:
# Mode may be far from actual central point.

# Mode only focuses on:
# Frequency
# not overall distribution.
```

```python
# Types of Mode:

# 1. Unimodal
# One mode.
# Example:
# 2, 3, 3, 5
# Mode = 3

# 2. Bimodal
# Two modes.
# Example:
# 2, 2, 5, 5, 8
# Modes = 2 and 5

# 3. Multimodal
# More than two modes.
# Example:
# 1,1,2,2,3,3
# Modes = 1,2,3

# 4. No Mode
# No repeated value.
# Example:
# 1,2,3,4,5
# No mode exists.
```

```python
x = ['Nokia','Samsung','Samsung','Apple','Oppo','Vivo']
statistics.mode(x)
```

```python
# Mode is heavily used in:
# - Categorical analysis
# - Market research
# - Recommendation systems
# - Survey analysis
# - Missing value imputation
```
