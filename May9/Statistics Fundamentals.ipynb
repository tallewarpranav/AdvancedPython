```python
# Measures of Central Tendency
# These measures help us identify the “center” or “typical value” of the data.

# Suppose a manager asks:
# “On average, how many products are sold daily?”
# or
# “What is the typical salary in the company?”
# or
# “What is the usual delivery time?”

# We need one value that summarizes the dataset.
# That central representative value is called: Central Tendency

# Suppose employee salaries are:
# 40k, 42k, 43k, 45k, 44k

# Most values are around 43k–44k.
# So center is around that region.
```

```python
# Mean
# Also called: Average

# The mean is calculated by:
# Adding all values and dividing by total number of values.

# Suppose 5 students scored:
# 70, 80, 90, 60, 100

# To find average marks:

# Step 1 → Add all marks
# 70 + 80 + 90 + 60 + 100 = 400

# Step 2 → Divide by number of students
# 400 ÷ 5 = 80

# So:
# Mean = 80
```

```python
# Mean can work with:

# Discrete Data
# Examples:
# - Number of defects
# - Number of orders

# Continuous Data
# Examples:
# - Height
# - Weight
# - Temperature
```

```python
# Mean is sensitive to outliers and skewed data.

# What Is an Outlier?
# An outlier is:
# An unusually large or small value.

# Example:
# Suppose salaries are:
# 40k, 42k, 43k, 44k, 10 crore

# Now average becomes extremely high.

# But most employees still earn around 40k.

# So mean becomes misleading.

# Real life example:
# Suppose company CEO salary is extremely high.
# Average company salary may look huge.
# But it does NOT represent normal employees.
```

```python
# Why This Matters in Data Science?

# Outliers can badly affect:
# - ML models
# - Business reports
# - Forecasting
# - Decision-making

# So sometimes:
# Median is preferred over mean.
```

```python
import statistics
statistics.mean([4,89,54,-7,-9,27,5])
```

```python
# What Is Median?

# Median means:
# The middle value of a dataset after arranging the data in sorted order.

# Median tells us:
# “What value lies exactly at the center?”

# Median is VERY useful when:
# - Outliers exist
# - Data is skewed
# - Extreme values distort averages

# Example:
# Suppose salaries are:
# 40k, 42k, 43k, 44k, 10 crore

# Mean Problem
# Average becomes extremely high because of CEO salary.
# But does that represent most employees?
# No.

# Median Solution

# Sorted salaries:
# 40k, 42k, 43k, 44k, 10 crore

# Middle value = 43k
# This better represents normal employees.

# Median is:
# Less sensitive to outliers.
# That is why it is heavily used in real business analytics.
```

| Case                  | Method                       |
| --------------------- | ---------------------------- |
| Odd number of values  | Exact middle value           |
| Even number of values | Average of two middle values |


```python
# CASE 1 — Odd Number of Values
# Dataset:
# 5, 76, 98, 32, 1, -6, 34, 3, -65

# Step 1 — Sort the Data
# Ascending order:
# -65, -6, 1, 3, 5, 32, 34, 76, 98

# Step 2 — Find Middle Value
# Total numbers = 9

# Middle position:
# (9 + 1)/2 = 5th position

# Step 3 — Identify 5th Value
# 5th value = 5

# So:
# Median = 5
```

```python
# CASE 2 — Even Number of Values
# Dataset:
# 5, 76, 98, 32, 1, -6, 34, 3, -65, 99

# Step 1 — Sort Data
# -65, -6, 1, 3, 5, 32, 34, 76, 98, 99

# Step 2 — Total Values
# Total = 10
# Even number of observations.
# So:
# No single middle value exists.

# Step 3 — Find Two Middle Values
# Middle values are:
# 5 and 32

# Step 4 — Take Average
# Formula:
# Median=
# (5 + 32) / 2

# Result:
# 18.5

# Final Answer
# Median = 18.5
```

```python
# Measures of Dispersion

# Q: Can two datasets have the same average but behave completely differently?
# Yes.

# That is why we need:
# Measures of Dispersion

# Dispersion means:
# How spread out the data values are.

# Also called:
# - Variability
# - Spread

# Dispersion tells us:
# “How far data points are from each other.”
```

```python
# Real-Life Example
# Suppose two factories produce products.

# Factory A Defects
# 10, 11, 10, 9, 10

# Factory B Defects
# 1, 20, 5, 25, 9

# Both factories may have similar averages.
# But:
# Factory A is stable.
# Factory B is highly inconsistent.
```

```python
# What Does Dispersion Reveal?

# Dispersion helps us understand:
# - Stability
# - Consistency
# - Reliability
# - Risk
# - Variation
```

```python
# Example

# Dataset A:
# 50, 50, 50, 50, 50
# Mean = 50

# Dataset B:
# 10, 20, 50, 80, 90
# Mean = 50

# Problem
# Both have same mean.
# But behavior is COMPLETELY different.

# Dataset A
# Very stable.
# Dataset B
# Highly spread out.

# This difference is captured using:
# Measures of Dispersion
```

```python
# Range: Difference between maximum and minimum value.

# Range tells us:
# “How wide the dataset is.”

# Real-Life Example — Student Marks

# Marks:
# 40, 50, 60, 80, 90

# Step 1 — Find Maximum
# Maximum = 90

# Step 2 — Find Minimum
# Minimum = 40

# Step 3 — Apply Formula
# 90 − 40 = 50

# Final Answer
# Range = 50

# What do we learn from range?
# Marks are spread across:
# 50 units.
```

```python
# Example — Manufacturing

# Machine temperatures:
# 70°C, 72°C, 75°C, 90°C

# Range:
# 90 − 70 = 20°C

# Why is this useful?
# Engineers quickly understand:
# - How much machine temperature fluctuates.
```

```python
# Example — Stock Market

# Stock prices:
# ₹100, ₹120, ₹180, ₹250

# Range:
# 250 − 100 = 150

# This tells investors:
# - How volatile the stock is.
```

```python
# Note:
# Range is sensitive to outliers.

# One extreme value can completely distort range.

# Example

# Dataset:
# 10, 12, 13, 14, 15

# Range:
# 15 − 10 = 5

# Now add one outlier:
# 10, 12, 13, 14, 1000

# Range:
# 1000 − 10 = 990

# Problem
# Most values are still around 10–15.
# But range suddenly becomes huge.
# So range becomes misleading.

# Why?
# Range uses ONLY:
# - Maximum value
# - Minimum value
# It ignores all middle data points.

# Example — Employee Salaries

# Suppose salaries:
# 40k, 42k, 43k, 45k, 5 crore

# Range becomes massive.
# But normal employee salaries are still around 40k.
# So range gives distorted picture.
```

```python
# Where Range Is Still Useful?

# Despite limitations, range is useful for:
# Quick understanding
# Initial exploration
# Comparing spread roughly
```

```python
# Example Dataset (Odd Number of Values)

# Suppose:
# 3, 7, 8, 5, 12, 14, 21, 13, 18

# Step 1 — Arrange in Ascending Order
# 3, 5, 7, 8, 12, 13, 14, 18, 21

# Total Values
# n = 9

# Finding Q2 (Median)
# Middle value:
# 5th value = 12

# So:
# Q2 = 12

# Now Divide Dataset Into Two Parts
# Exclude median while dividing.

# Lower Half
# 3, 5, 7, 8

# Upper Half
# 13, 14, 18, 21

# Finding Q1
# Median of lower half:
# 3, 5, 7, 8

# Middle values:
# 5 and 7

# Average:
# (5+7)/2	​
# =6

# So:
# Q1 = 6

# Finding Q3
# Upper half:
# 13, 14, 18, 21

# Middle values:
# 14 and 18

# Average:
# (14 + 18) / 2
# =16

# So:
# Q3 = 16
```

```python
# Interquartile Range (IQR)
# IQR measures the spread of the middle 50% of the data.

# Formula:
# IQR = Q3 − Q1

# |----25%----|====Middle 50%====|----25%----|
#       Q1             Q2               Q3
# IQR measures ONLY the highlighted middle region.
```

```python
# IQR helps us understand:
# - Actual spread of normal data
# - Variability
# - Outliers
# - Skewness

# without getting heavily affected by extreme values.
```

```python
# Example — Employee Salaries

# Suppose salaries are:
# 30k, 35k, 40k, 42k, 45k, 50k, 55k, 60k, 5 crore

# Problem with Range
# Range becomes massive because of:
# 5 crore

# But most employees earn around:
# 30k–60k

# IQR Solution
# IQR ignores extreme ends and focuses on:
# Middle 50%
# So it gives much more realistic understanding.

# Step-by-Step Example

# Dataset:
# 2, 4, 6, 8, 10, 12, 14, 16, 18

# Step 1 — Find Q2 (Median)
# Middle value:
# 10

# Step 2 — Lower Half
# 2, 4, 6, 8
# Q1:
# Middle values = 4 and 6
# Q1
# ​(4+6)/2
# =5

# Step 3 — Upper Half
# 12, 14, 16, 18
# Q3:
# Middle values = 14 and 16
# Q3
# (14+16)/2
# =15

# Step 4 — Calculate IQR
# IQR = 15−5 = 10

# Final Answer
# IQR = 10

# What does this mean?
# Middle 50% data is spread across:
# 10 units.
```

| Range                       | IQR                     |
| --------------------------- | ----------------------- |
| Uses extreme values         | Uses middle values      |
| Highly affected by outliers | Robust against outliers |
| Less stable                 | More reliable           |


```python
# Example — House Prices

# Suppose city has:
# Normal homes = 50L–1Cr
# Few luxury villas = 50Cr

# Range becomes huge.

# IQR still captures normal market spread properly.
# That is why analysts often prefer IQR.
```

```python
# Outlier Rule

# A value is considered outlier if:
# Lower Bound: Q1 − 1.5 × IQR
# Upper Bound: Q3 + 1.5 × IQR

# Real-Life Example

# Suppose:
# Q1 = 20
# Q3 = 40

# IQR:
# 40 − 20 = 20

# Lower Limit
# 20 − 1.5(20) = −10

# Upper Limit
# 40 + 1.5(20) = 70

# Interpretation
# Any value:
# Below −10
# or
# Above 70
# is treated as outlier.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")

sns.histplot(df["price"], kde=True)
plt.axvline(df["price"].mean(), color='red', linestyle='--', label='Mean')
plt.axvline(df["price"].median(), color='green', linestyle='-', label='Median')
plt.axvline(df["price"].quantile(0.25), color='blue', linestyle=':', label='Q1')
plt.axvline(df["price"].quantile(0.75), color='orange', linestyle='-.', label='Q3')
plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# lower_bound = Q1 - 1.5 * IQR
# upper_bound = Q3 + 1.5 * IQR
# IQR = Q3 - Q1

q1 = df["price"].quantile(0.25)
q3 = df["price"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")

sns.histplot(df["price"], kde=True)
plt.axvline(lower_bound, color='purple', linestyle='--', label='Lower Bound')
plt.axvline(upper_bound, color='purple', linestyle='--', label='Upper Bound')
plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# Lets plot boxplot to identify outliers in price column
sns.boxplot(x=df["price"])
plt.axvline(lower_bound, color='red', linestyle='--', label='Lower Bound')
plt.axvline(upper_bound, color='red', linestyle='--', label='Upper Bound')
plt.title("Boxplot of Car Prices")
plt.xlabel("Price")
plt.show()
```

```python
# Remove outliers from price column
df_no_outliers = df[ (df["price"] >= lower_bound) & (df["price"] <= upper_bound) ]
print("Original Data Shape:", df.shape)
print("Data Shape after Removing Outliers:", df_no_outliers.shape)
```

```python
sns.histplot(df_no_outliers["price"], kde=True)
plt.title("Distribution of Car Prices (Outliers Removed)")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
# ------------------------------------------------
# TASK 12 - FIND QUARTILES
# ------------------------------------------------
# Dataset:
# 3, 5, 7, 8, 12, 13, 14, 18, 21
#
# Questions:
# 1. Find Q1
# 2. Find Q2
# 3. Find Q3
# 4. Verify using numpy.percentile()
print("12.Find Quartiles:")
data = [3, 5, 7, 8, 12, 13, 14, 18, 21]
# Step 1: Arrange in ascending order (already sorted)
# Step 2: Find Q2 (Median)
n = len(data)
q2 = data[n // 2]  # Middle value for odd number of values
print("Q2 (Median):", q2)
# Step 3: Find Q1
lower_half = data[:n // 2]  # Lower half excluding median
n_lower = len(lower_half)
q1 = lower_half[n_lower // 2]  # Middle value of lower half
print("Q1:", q1)
# Step 4: Find Q3
upper_half = data[n // 2 + 1:]  # Upper half excluding median
n_upper = len(upper_half)
q3 = upper_half[n_upper // 2]  # Middle value of upper half
print("Q3:", q3)
# Step 5: Verify using numpy.percentile()
import numpy as np
q1_np = np.percentile(data, 25)
q2_np = np.percentile(data, 50)
q3_np = np.percentile(data, 75)
print("Q1 (NumPy):", q1_np)
print("Q2 (NumPy):", q2_np)
print("Q3 (NumPy):", q3_np)

```

```python

# ------------------------------------------------
# TASK 13 - CALCULATE IQR
# ------------------------------------------------
# Using the quartiles from TASK 12:
#
# Questions:
# 1. Calculate Interquartile Range (IQR)
# 2. What does IQR represent?
# 3. Why is IQR more reliable than range?
print("13.Calculate IQR:")
# Assuming Q1 and Q3 from Task 12
q1 = 7
q3 = 14
# Step 1: Calculate IQR
iqr = q3 - q1
print("IQR:", iqr)
# Step 2: What does IQR represent?
print("IQR represents the spread of the middle 50% of the data, indicating how much the central portion of the data varies.")
# Step 3: Why is IQR more reliable than range?
print("IQR is more reliable than range because it is not affected by outliers or extreme values, " \
"as it focuses only on the middle 50% of the data, while range can be significantly distorted by outliers.")
```

```python
# ------------------------------------------------
# TASK 14 - IDENTIFY OUTLIERS USING IQR
# ------------------------------------------------
# Dataset:
# 5, 7, 8, 10, 12, 15, 18, 20, 100
#
# Questions:
# 1. Find Q1 and Q3
# 2. Find IQR
# 3. Calculate lower and upper bounds
# 4. Identify outliers
print("14.Identify Outliers Using IQR:")
data = [5, 7, 8, 10, 12, 15, 18, 20, 100]
# Step 1: Find Q1 and Q3
import numpy as np
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
print("Q1:", q1)
print("Q3:", q3)
# Step 2: Find IQR
iqr = q3 - q1
print("IQR:", iqr)
# Step 3: Calculate lower and upper bounds
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
# Step 4: Identify outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```

```python

# ------------------------------------------------
# TASK 18 - PYTHON IMPLEMENTATION TASK
# ------------------------------------------------
# Create a Python list containing:
# 10 employee salaries.
#
# Using Python:
# 1. Find mean
# 2. Find median
# 3. Find mode
# 4. Find range
# 5. Find Q1 and Q3
# 6. Find IQR
print("18.Python Implementation Task:")
salaries = [30000, 35000, 40000, 45000, 50000, 55000, 60000, 65000, 70000, 75000]
mean_salary = statistics.mean(salaries)
median_salary = statistics.median(salaries)
mode_salary = statistics.mode(salaries)  # Note: This will raise an error if there is no mode
range_salary = max(salaries) - min(salaries)
q1_salary = np.percentile(salaries, 25)
q3_salary = np.percentile(salaries, 75)
iqr_salary = q3_salary - q1_salary
print("Mean Salary:", mean_salary)
print("Median Salary:", median_salary)
print("Mode Salary:", mode_salary)
print("Range of Salaries:", range_salary)
print("Q1 Salary:", q1_salary)
print("Q3 Salary:", q3_salary)
print("IQR of Salaries:", iqr_salary)
#find outliers using IQR
lower_bound = q1_salary - 1.5 * iqr_salary
upper_bound = q3_salary + 1.5 * iqr_salary
outliers = [x for x in salaries if x < lower_bound or x > upper_bound]
print("Outliers in Salaries:", outliers)

```

```python
# ------------------------------------------------
# TASK 20 - MINI CASE STUDY
# ------------------------------------------------
# A food delivery company observed the following delivery times:
#
# 20, 22, 25, 24, 21, 23, 120
#
# Questions:
# 1. Calculate mean and median.
# 2. Which measure better represents delivery performance?
# 3. Why?
# 4. Identify whether any outlier exists.
# ================================================================
print("20.Mini Case Study:")
delivery_times = [20, 22, 25, 24, 21, 23, 120]
mean_delivery = statistics.mean(delivery_times)
median_delivery = statistics.median(delivery_times)
print("Mean Delivery Time:", mean_delivery)
print("Median Delivery Time:", median_delivery)
print("The median better represents delivery performance because it is not affected by the outlier (120), " \
"while the mean is significantly increased due to the outlier, giving a misleading impression of the typical delivery time.")
print("Yes, there is an outlier (120) in the delivery times, which is " \
"much higher than the other values and can distort the analysis of delivery performance.")
```
